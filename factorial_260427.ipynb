countdown_num() 함수 : 카운트 다운 횟수를 반환한다


```python
def countdown_num(n):
  if n <= 0:
    return 0
  else:
    print(n, '이(가) 양수인 경우!')
    result = countdown_num(n-1) + 1
    return result

print('반환값:', countdown_num(0))
```

```python
print('반환값:', countdown_num(1))
```

n = 1이면 재귀 호출이 한 번 발생하며 countdown_num(0)은 0을 반환하며 종료된다. 따라서 result = 0 + 1이 실행되어 최종적으로 1을 반환한다.


```python
print('반환값:', countdown_num(2))
```

```python
print('반환값:', countdown_num(3))
```

계승

음이 아닌 정수 n이 주어졌을 때 1부터 n까지의 곱을 n의 계승 또는 팩토리얼이라 하고 n!로 표시한다

- 기저조건은 n == 0이고, 1을 반환한다

- n이 0보다 크면 (n-1)의 계승과 n의 곱함 값을 반환한다


```python
def factorial(n):
  if n == 0: # 기저조건
    return 1

  else:
    recursion_step = factorial(n-1) # (n-1)의 계승
    result = n * recursion_step
    return result
```

콜라츠 추측

- 기저조건 num == 1

- num % 2 == 0: 짝수인 경우. 2로 나누기

- 기타else : 홀수 인 경우. 3배 더하기 1

```python
def collatz(num):
  if num == 1: # 기저조건
    print(1)
  elif num%2 == 0: # 짝수인 경우
    print(num, end=' -> ')
    collatz(num//2)
  else: # 홀수인 경우
    print(num, end=' -> ')
    collatz(num*3 + 1)
```

```python
collatz(7)
```

```python
def collatz_count(num):
  if num == 1: # 기저조건
    return 0
  elif num % 2 == 0: # 짝수인 경우
    return collatz_count(num//2) + 1
  else: #홀수인 경우
    return collatz_count(num*3 +1) +1
```

DP 상향식과 하향식 차이

상향식 Bottom-Up

- 작은 문제부터 차례대로 계산
- 보통 for문 사용
- 계산한 값을 dp 리스트에 저장
- 재귀 호출 없음

```python
# 상향식 Bottom-up 계단 오르기

def climb_bottom_up(n):
    dp = [0] * (n + 1)
    dp[1] = 1
    if n >= 2:
        dp[2] = 2

    for i in range(3, n + 1):
        dp[i] = dp[i-1] + dp[i-2]

    return dp[n]
```

하향식 Top-Down

- 큰 문제에서 시작해서 작은 문제로 쪼갬
- 보통 재귀함수 사용
- 이미 계산한 값은 memo에 저장
- 같은 계산을 다시 하지 않음

```python
# 하향식 : n칸까지 몇 가지 방법으로 올라갈 수 있는가

def climb_top_down(n, memo):
    if n <= 2:
        return n

    if memo[n] != 0:
        return memo[n]

    memo[n] = climb_top_down(n-1, memo) + climb_top_down(n-2, memo)
    return memo[n]

n = 5
memo = [0] * (n + 1)
print(climb_top_down(n, memo))
```

Singly Linked-List


```python
# 단일 링크드 리스트
class SLinkedList:

    # S_L_list에서 쓸 노드
    class Node:
        def __init__(self, v, n = None):
            self.value = v  # 저장된 데이터
            self.next = n   # 다음 노드 가리키는 변수


    # S_L_list에서 필요한 변수
    def __init__(self):
        self.head = None  # 첫 생성시 내부에는 노드가 없으므로


    # 삽입
    def insertNode(self, v):  # 추가할 데이터
        # 만일 처음 노드일 경우 -> head값이 None임
        if self.head is None:
            # head에 새 노드를 저장
            self.head = self.Node(v)
        else:  # 이미 노드가 있는 경우
            # head에 새 노드를 저장
            # 기존 head에 저장된 노드는 새로 생성할 노드의 next로 저장
            self.head = self.Node(v, self.head)


    def printNode(self):
        # 데이터가 없을 때
        if self.head is None:
            print("저장된 데이터가 없음")
            return
        else:
            print("<현재 리스트 구조>", end='\t')
            link = self.head  # 처음은 head를 지정

            # link가 가리키는 노드가 없을 때까지 반복
            while link:
                print(link.value, '->', end=' ')
                link = link.next  # 다음 노드로 이동

            print()  # 줄바꿈


    # 삭제
    def deleteNode(self):
        # 노드가 없으면 skip
        if self.head is None:
            print("삭제할 노드가 없습니다.")
            return
        else:
            # head를 다음 노드로 변경
            self.head = self.head.next

    # 탐색
    def searchNode(self, v):
        if self.head is None:
            return "저장된 데이터가 없음"

        link = self.head
        position = 1
        while link:
            if link.value == v:
                return f"{v}는 {position}번째에 있습니다."
            link = link.next
            position += 1
        return f"{v}를 찾을 수 없습니다."
```

```python
s = SLinkedList()
s.insertNode(10)
s.insertNode(20)
s.insertNode(30)
s.printNode()
```

Doubly Linked-List

```python
# Doubly Linked List 예제
# 기능: head 삽입/삭제, tail 삽입/삭제, 출력, 탐색

class DoublyLinkedList:

    class Node:
        def __init__(self, v):
            self.value = v
            self.prev = None
            self.next = None

    def __init__(self):
        self.head = None
        self.tail = None

    # head 기준 삽입
    def insert_head(self, v):
        new_node = self.Node(v)

        if self.head is None:
            self.head = new_node
            self.tail = new_node
        else:
            new_node.next = self.head
            self.head.prev = new_node
            self.head = new_node

    # tail 기준 삽입
    def insert_tail(self, v):
        new_node = self.Node(v)

        if self.tail is None:
            self.head = new_node
            self.tail = new_node
        else:
            new_node.prev = self.tail
            self.tail.next = new_node
            self.tail = new_node

    # head 기준 삭제
    def delete_head(self):
        if self.head is None:
            print("삭제할 노드가 없습니다.")
            return

        if self.head == self.tail:
            self.head = None
            self.tail = None
        else:
            self.head = self.head.next
            self.head.prev = None

    # tail 기준 삭제
    def delete_tail(self):
        if self.tail is None:
            print("삭제할 노드가 없습니다.")
            return

        if self.head == self.tail:
            self.head = None
            self.tail = None
        else:
            self.tail = self.tail.prev
            self.tail.next = None

    # 출력: head → tail
    def print_from_head(self):
        if self.head is None:
            print("저장된 데이터가 없습니다.")
            return

        cur = self.head
        print("head -> tail:", end=" ")
        while cur is not None:
            print(cur.value, end=" <-> ")
            cur = cur.next
        print("None")

    # 출력: tail → head
    def print_from_tail(self):
        if self.tail is None:
            print("저장된 데이터가 없습니다.")
            return

        cur = self.tail
        print("tail -> head:", end=" ")
        while cur is not None:
            print(cur.value, end=" <-> ")
            cur = cur.prev
        print("None")

    # 탐색
    def search(self, v):
        cur = self.head

        while cur is not None:
            if cur.value == v:
                return True
            cur = cur.next

        return False
```

```python
# 실행 예제
dll = DoublyLinkedList()

print("1. 저장된 노드가 없을 때 삭제")
dll.delete_head()
dll.delete_tail()

print("\n2. head 기준 삽입")
dll.insert_head(10)
dll.insert_head(20)
dll.insert_head(30)
dll.print_from_head()
dll.print_from_tail()

print("\n3. tail 기준 삽입")
dll.insert_tail(40)
dll.insert_tail(50)
dll.print_from_head()
dll.print_from_tail()

print("\n4. 탐색")
print("20 탐색:", dll.search(20))
print("99 탐색:", dll.search(99))

print("\n5. head 기준 삭제")
dll.delete_head()
dll.print_from_head()

print("\n6. tail 기준 삭제")
dll.delete_tail()
dll.print_from_head()

print("\n7. 남은 데이터 확인")
dll.print_from_head()
dll.print_from_tail()
```

head는 앞, tail은 뒤. Doubly Linked List는 next뿐 아니라 prev도 있어서 양방향 이동이 가능하다
