**Node Class**

```python
class Node:
  def __init__(self, data):
    self.data = data # data는 값을 가리키는 변수 (속성, attribute)
    self.next = None # next는 다음 노드를 가리키는 변수
```

연결리스트(Linked List) 자체를 클래스로 만들기

방법 1 : 변수로 각각 만들고 연결하기

```python
# 기차 칸 3개 만들기
car1 = Node(10)
car2 = Node(20)
car3 = Node(30)

# 연결하기
car1.next = car2
car2.next = car3

# 기차 칸 처음부터 끝까지 달려보기
gogocar = car1

while gogocar is not None:
  print(gogocar.data)
  gogocar = gogocar.next
```

방법 2 : head로 바로 연결하기 (중간 변수 생략)

```python
head = Node(10)
head.next = Node(20)
head.next.next = Node(30)

current = head
while current is not None :
    print(current.data)
    current = current.next

# 출력 : 10 -> 20 -> 30
```

- head부터 마지막 노드까지 이동하면서 data를 출력한다
- 이 때 head는 연결리스트의 시작이므로 head가 이동하면 연결리스트를 잃게된다.
- 마지막 노드의 next는 None을 가리키므로, 노드가 None이 아닐 동안 계속 이동하면서 data를 출력한다.

- __init__ : Node를 만들 때 자동으로 실행되는 함수. (쉽게 말하면 공장의 설명서)

```python
class Node:
  def __init__(self, data):
    self.data = data
    self.next = None

head = Node(1)
head.next = Node(2)
head.next.next = Node(3)


# 끝에 4 추가하기
node = head
while node:
  if node.next is None:
      node.next = Node(4)
      break
  node = node.next # break랑 같은 레벨

# 확인 출력
node = head
while node :
  print(node.data, end = " ")
  node = node.next

# 출력 : 1 2 3 4
```

끝에 노드 추가

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

head = Node(1)
head.next = Node(2)
head.next.next = Node(3)

# 끝에 4 추가하기
node = head
while node.next:
    node = node.next
node.next = Node(4)


# 처음에 0 추가하기
node = Node(0)
node.next = head
head = node


# 확인 출력
node = head
while node:
  print(node.data, end = " ")
  node = node.next
```

node = Node(0)   →  [0] 만들기

node.next = head →  [0] → [1] 연결

head = node      →  head를 [0]으로 옮기기

결과: head → [0] → [1] → [2] → [3] → [4] → None

**Stack**

stack 클래스 만들기 (프린글스 통 비유)

```python
class Node: # 과자 한 개 설계도
    def __init__(self, data): # 과자 만들 때
        self.data = data # 과자 맛 저장 (1,2,3...)
        self.next = None # 아래 과자 연결 (처음엔 없음)

class Stack: # 프링글스 통 설계도
    def __init__(self):  # 빈 통 만들 때
        self.top = None # 아직 과자 없음

    def push(self, data): # 과자넣기
        node = Node(data) # 새 과자 만들기
        node.next = self.top # 새 과자 밑에 기존 과자 연결
        self.top = node # top을 새 과자로 변경

    def pop(self): # 과자 꺼내기
        if self.top is None: # 만약 통이 비었으면?
            return None # 꺼낼 과자가 없다
        node = self.top # 제일 위 과자 기억하두고
        self.top = self.top.next # top 아래로 변경
        return node.data # 꺼낸 과자 반환

    def peek(self): # 훔쳐보기
        if self.top is None: # 비었으면
            return None # 볼게 없다
        return self.top.data # 제일 위 과자만 확인 / 꺼내지 않는다.

    def is_empty(self): # 통이 비었는가
        return self.top is None # 비었으면 True, 아니면 False




s = Stack() # 프링그스 통 만들기
s.push(1) # 1 넣기
s.push(2) # 2 넣기
s.push(3) # 3 넣기

print(s.peek())     # 3
print(s.pop())      # 3
print(s.pop())      # 2
print(s.pop())      # 1
print(s.is_empty()) # True
```

Stack의 역순 : 넣은 순서 반대로 꺼내는 것

주어진 문자열을 역순으로 반환하는 함수를 작성하라

입력 : s = "aircraft"

출력 : "tfarcria"

```python
# 앞서 구현한 Stack 클래스를 이용한다

def reverse_word(word: str) -> str:
  answer: str = ""
  s = Stack()

  for w in word:
    s.push(w)
  while not s.is_empty():
    answer += s.pop()

  return answer


# 테스트 코드
print(reverse_word("aircraft"))
```

괄호의 짝이 바르면 Ture, 바르지 않으면 False 반환하는 함수를 작성하라

예를들어 ((a*(b+c))-d) / e는 괄호의 짝이 올바르지만, (((a*(b+c))-d) / e는 괄호의 짝이 맞지 않는다. 괄호는 소괄호 ()만 사용한다.

```python
# 짝 찾기
def check_bracket(expression: str) -> bool:
  s = Stack()
  for exp in expression:
    if exp == "(":
      s.push(exp)
    elif exp == ")":
      if not s.pop():
        return False
  return True if s.is_empty() else False


# 테스트 코드
print(check_bracket("((a * ( b + c )) -  d) / e"))
print(check_bracket("(((a * ( b + c )) -  d) / e"))
```

응용ver. 소괄호, 중괄호, 대괄호의 짝이 맞는지 검사하기

```python
def check_bracket2(expression: str) -> bool:
  brackets : dict[str, str] = {")": "(","}": "{", "]": "["} # Corrected the mapping for ']' from '[]' to '['
  s = Stack()

  for exp in expression:
    if exp in brackets.values(): # If it's an opening bracket
      s.push(exp)
    elif exp in brackets: # If it's a closing bracket
      pop_bracket = s.pop() # Pop the top element from the stack
      if not pop_bracket or pop_bracket != brackets[exp]: # Check if stack is empty or if popped bracket doesn't match
        return False
  return True if s.is_empty() else False


# 테스트 코드
print(check_bracket2("[{a * (b + c)} - d] / e"))
print(check_bracket2("[{a * (b + c)] - d] / e"))
```

우리가 평소에 쓰는 수식은 '중위 표기법'이다

후위 표기법은 연산자를 뒤에 쓴다  (컴퓨터가 계산하기 훨씬 편한 방식)


스택을 쓰는 이유 : 우선순위

```python
# 직접 구현한 스택을 이용한 함수

def to_postfix(expresstion: str) -> str:
  op: dict[str, int] = {"+": 1, "-": 1, "*": 2, "/": 2}
  res: str = ""
  s = Stack()

  for exp in expresstion:
    if exp.isnumeric():
      res += exp

    elif exp in op:
      if not s.is_empty() and (op[exp] <= op[s.peek()]):
        res += s.pop()
      s.push(exp)

  while not s.is_empty():
      res += s.pop()
  return res  # ← while 밖!


# 테스트 코드
for expr in ("3 + 5 * 2", "3 * 5 + 2"):
  print(f"{expr} -> {to_postfix(expr)}")
```
