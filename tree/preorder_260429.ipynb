방문 결과를 리스트로 반환하기 #1

방문한 노드를 리스트로 반환하는 함수 만들기
- 변수 res를 만든다
- 방문처리는 배열의 값을 리스트로 만들어 변수 res에 할당하고 재귀호출한 결과를 res에 더한다

```python
def preorder(tree, i=0):
  if i < len(tree):
    res = [tree[i]]
    left = 2 * i + 1
    right = 2 * i + 2
    if left < len(tree) and tree[left] != None:
      res += preorder(tree, left)
    if right < len(tree) and tree[right] != None:
      res += preorder(tree, right)
    return res
```

```python
# 테스트 코드
tree = ["A", "B", "C", "D", "E", "F", None, "G"]
preorder(tree)
```

방문 결과를 리스트로 반환하기 #3

파이썬은 함수 안에 함수를 정의할 수 있으므로 위의 함수를 다른 함수 안에 넣어보자

```python
def preorder(tree):
  def _preorder(i=0):
    if i < len(tree):
      res.append(tree[i])
      left = 2 * i + 1
      right = left + 1
      if left < len(tree) and tree[left] is not None:
        _preorder(left)
      if right < len(tree) and tree[right] is not None:
        _preorder(right)
      return res

  res = []
  _preorder()
  return res


# 테스트 코드
tree = ["A", "B", "C", "D", "E", "F", None, "G"]
print(preorder(tree))
```

preorder(tree)처럼 매개변수에 tree만 넣어주면 되고, 정상적으로 작동한다.

 중위 순회 함수와 후위 순회 함수는 res.append(tree[i])의 위치만 수정하면 되니까 좋은 듯하다.

```python
def preorder(tree):
    def _preorder(i = 0):
        if i >= len(tree) or tree[i] is None: #인덱스를 벗어나거나, 원소가 None이면 종료
            return
        res.append(tree[i])
        _preorder(2 * i + 1)
        _preorder(2 * i + 2)

    res = []
    _preorder()
    return res
```

중위 순회 함수 만들기


- 이미 만든 전위 순회 함수의 코드에서 자신을 재귀호출하는 순서만 바꾸면 된다.

-  중위 순회는 왼쪽 노드 → 현재(부모) 노드 → 오른쪽 노드 순으로 방문


```python
def inorder(tree):
  def _inorder(i=0):
    if i >= len(tree) or tree[i] is None:
      return
    _inorder(2 * i + 1)
    res.append(tree[i])
    _inorder(2 * i + 2)

  res = []
  _inorder()
  return res


# 테스트 코드
tree = ["A", "B", "C", "D", "E", "F", None, "G"]
print(inorder(tree))
```

후회 선회 함수 만들기

- 방문하는 코드의 위치 바꾸기
- 재귀 호출하여 왼쪽 노드 - 오른쪽 모드 - 현재 노드

```python
def postorder(tree):
  def _postorder(i=0):
    if i >= len(tree) or tree[i] is None:
      return
    _postorder(2 * i + 1)
    _postorder(2 * i + 2)
    res.append(tree[i])


  res = []
  _postorder()
  return res



# 테스트 코드
tree = ["A", "B", "C", "D", "E", "F", None, "G"]
print(postorder(tree))
```
