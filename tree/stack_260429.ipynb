스택으로 순회함수 만들기


```python
def preorder(tree):
  if not tree:
      return []
  res, stack = [], [0]

  while stack:
      parent = stack.pop()
      res.append(tree[parent])
      left = 2 * parent + 1
      right = left + 1
      if right < len(tree) and tree[right] is not None:
          stack.append(right)
      if left < len(tree) and tree[left] is not None:
          stack.append(left)
  return res
```

코드를 여러 번 읽다 보면, parent, left, right 세 개의 변수를 쓰지 않고도 index 하나로도 구현할 수 있음을 알 수 있다.


- index라는 변수를 만들고 부모 노드의 인덱스를 할당한 후 방문 처리한다.
- 여기에 2를 곱한 후에 2를 더하여 오른쪽 노드의 인덱스를 구하여 방문 처리한다.
- 다시 1을 빼서 왼쪽 노드의 인덱스를 구하여 방문 처리한다.

```python
def preorder(tree):
  if not tree:
    return []
  res, stack = [], [0]


  while stack:
    index = stack.pop()
    res.append(tree[index])
    index = 2 * index + 2
    if index < len(tree) and tree[index] is not None:
      stack.append(index)
    index -= 1
    if index < len(tree) and tree[index] is not None:
      stack.append(index)

  return res


tree = ["A", "B", "C", "D", "E", "F", None, "G"]
print(preorder(tree))
```

스택 중위순회

- 현재의 노드의 왼쪽에 노드가 있다면 왼쪽 노드부터 방문
- 루트 노드부터 가장 왼쪽 리프 노드가 나올 때까지 차례로 스택에 저장해서 방문을 보류한다.



```python
def inorder(tree):
  if not tree:
    return []
  index = 0
  res, stack = [], []


  while True:
    if index < len(tree) and tree[index] is not None:
      stack.append(index)
      index = 2 * index + 1

    elif stack:
      index = stack.pop()
      res.append(tree[index])
      index = 2 * index + 2

    else:
      break

  return res


#테스트 코드
tree = ["A", "B", "C", "D", "E", "F", None, "G"]
print(inorder(tree))
```

스택을 이용한 후위 순회


- 스택 2에 넣는 순서가 전위 순회의 순서와 반대

```python
def postorder(tree):
    if not tree:
        return []
    res, stack = [], [0]
    visit_order = []

    while stack:
        index = stack.pop()
        visit_order.append(index)
        index = 2 * index + 1
        if index < len(tree) and tree[index] is not None:
            stack.append(index)
        index = index + 1
        if index < len(tree) and tree[index] is not None:
            stack.append(index)

    while visit_order:
        index = visit_order.pop()
        res.append(tree[index])
    return res


#테스트 코드
tree = ["A", "B", "C", "D", "E", "F", None, "G"]
print(postorder(tree))
```

이진트리(Binary Tree) 문제 풀기 1

이진트리와 키가 주어질 때, 이진트리의 레벨 순서에서 가장 먼저 사용할 수 있는 위치에 키를 삽입하는 함수를 만들어라.

해석 : 빈 자리 중에서 가장 먼저 발견되는 곳에 값을 넣어라 (레벨 순서 채우기)

```python
from collections import deque

class Node:
  def __init__(self, data):
    self.data = data
    self.left = None
    self.right = None


def insert_lavel_order(root, key):
  new_node = Node(key)


  # 트리가 이미 비어있으면 바로 넣기
  if not root:
    return new_node

  q = deque([root])


  while q:
    node = q.popleft()


    # 왼쪽 비여있으면 여기 넣고 끝
    if not node.left:
      node.left = new_node
      return root
    else:
      q.append(node.left)

    # 오른쪽 비어있으면 여기 넣고 끝
    if not node.right:
      node.right = new_node
      return root
    else:
      q.append(node.right)


  # 반복문 끝나고 실행되게 위치 이동
  # level_order_print(root) # This line was causing an issue as it was always called outside the loop. Removed for now.
  return root

def level_order_print(root):
    if not root:
        return
    q = deque([root])
    while q:
        node = q.popleft()
        print(node.data, end=' ')
        if node.left:
            q.append(node.left)
        if node.right:
            q.append(node.right)
    print()

# 테스트 코드
print("트리 초기화 후: ")
root_node = Node("A")
level_order_print(root_node)

print("\nB 삽입 후: ")
insert_lavel_order(root_node, "B")
level_order_print(root_node)

print("\nC 삽입 후: ")
insert_lavel_order(root_node, "C")
level_order_print(root_node)

print("\nD 삽입 후: ")
insert_lavel_order(root_node, "D")
level_order_print(root_node)

print("\nE 삽입 후: ")
insert_lavel_order(root_node, "E")
level_order_print(root_node)

print("\nF 삽입 후: ")
insert_lavel_order(root_node, "F")
level_order_print(root_node)

print("\nG 삽입 후: ")
insert_lavel_order(root_node, "G")
level_order_print(root_node)
```

- 큐에 넣고 하나씩 꺼냄
- 왼쪽 → 오른쪽 순서로 빈 자리 찾기
- 처음 빈 자리 발견하면 바로 삽입하고 종료

이진트리(Binary Tree) 문제 풀기 2

크기가 n인 이진트리가 주어질 때, 주어진 값(key)를 삭제하고, 레벨 순서상 가장 마지막 값으로 대체하는 함수를 만들어라


```python
from collections import deque

class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

def level_order_print(root):
    if not root:
        print("비어있는 트리")
        return

    q = deque([root])

    while q:
        node = q.popleft()
        print(node.data, end=" ")

        if node.left:
            q.append(node.left)
        if node.right:
            q.append(node.right)

    print()

def delete_deepest(root, deepest_node):
    q = deque([root])

    while q:
        node = q.popleft()

        if node.left:
            if node.left == deepest_node:
                node.left = None
                return
            else:
                q.append(node.left)

        if node.right:
            if node.right == deepest_node:
                node.right = None
                return
            else:
                q.append(node.right)

def delete_node(root, key):
    if not root:
        return None

    if not root.left and not root.right:
        if root.data == key:
            return None
        return root

    q = deque([root])
    key_node = None
    last_node = None

    while q:
        last_node = q.popleft()

        if last_node.data == key:
            key_node = last_node

        if last_node.left:
            q.append(last_node.left)

        if last_node.right:
            q.append(last_node.right)

    if key_node:
        key_node.data = last_node.data
        delete_deepest(root, last_node)

    return root

# 테스트 트리 만들기
root = Node(13)
root.left = Node(12)
root.right = Node(10)
root.left.left = Node(4)
root.left.right = Node(19)
root.right.left = Node(16)
root.right.right = Node(9)

print("삭제 전:")
level_order_print(root)

root = delete_node(root, 12)

print("12 삭제 후:")
level_order_print(root)
```

12를 지우는 게 아니라, 12 자리에 가장 마지막 노드 9를 넣고 원래 9 노드를 삭제하는 방식이다.
